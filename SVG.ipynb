````python
import subprocess
import time

# Start the server and redirect logs to a file
with open("ollama_server.log", "w") as f:
    subprocess.Popen(["ollama", "serve"], stdout=f, stderr=f)

# Give the server a few seconds to wake up
time.sleep(5)
print("🚀 Ollama server is running in the background!")
````

````python
!pip install ollama
import ollama
from IPython.display import SVG, display, HTML

def generate_colab_svg():
    movie = input("Enter a movie/song: ")

    response = ollama.chat(model='llama3.2', messages=[
        {'role': 'system', 'content': 'You are an SVG designer. Output ONLY raw SVG code.'},
        {'role': 'user', 'content': f'Draw a minimalist icon for: {movie}'}
    ])

    svg_code = response['message']['content'].strip()
    svg_code = svg_code.replace("```svg", "").replace("```", "").strip()

    display(HTML(f'<div style="background:white; padding:20px;">{svg_code}</div>'))

generate_colab_svg()
````
